# R Programming - Loops, Functions, Sequences

## For Loop
`for (name in vector) { commands }` runs the commands once for each value in `vector`, with `name` taking that value each time.

```r
for (i in 1:5) {
    print(i^2)
}
```

The vector can be any set of values, such as `c(2,4,6,7)`, not just a range like `1:5`.

```r
for (i in c(2,4,6,7)) {
    print(i^2)
}
```

A `for` loop can be placed inside a function to accumulate a result, such as a counter, across all elements of a vector.

```r
x = c(2,4,6,8,10,12)

excount = function(x) {
    count = 0
    for (xval in x) {
        if (xval/2 > 3)
            count = count+1
    }
    print(count)
}

excount(x)
```

## Nested For Loop
A loop can contain another loop inside it. The inner loop completes fully for each single value of the outer loop.

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
    for (y in sweet) {
        print(paste(x, y))
    }
}
```

## The break Command
`break` immediately stops the loop and exits it, skipping any remaining values in the vector.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
    if (x == "tea") {
        break
    }
    print(x)
}
```

## The next Command
`next` skips the rest of the current iteration and moves on to the next value, without ending the loop.

```r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
    if (x == "lemonade") {
        next
    }
    print(x)
}
```

## While Loop
`while (condition) { commands }` repeats the commands as long as `condition` stays `TRUE`. It is used when the number of iterations is not known in advance.

```r
i = 1

while (i < 10) {
    print(i^2)
    i = i+2
}
```

A `while` loop is useful when the stopping point depends on user input or a computed value rather than a fixed count.

```r
sumfunction = function() {
    sum = 0
    number = as.integer(readline(prompt="Please select any number less than 25: "))
    while (number <= 25) {
        sum = sum + number
        number = number + 1
    }
    print(paste("The sum of numbers received from the While Loop: ", sum))
}

sumfunction()
```

## Repeat Loop
`repeat { commands }` runs the commands indefinitely; there is no built-in condition, so a `break` statement must be used to stop it.

```r
i = 1

repeat {
    print(i^2)
    i = i+2
    if (i > 10)
        break
}
```

## Repeat Loop with next
Inside a `repeat` loop, `next` returns control to the start of the loop body, while `break` still ends the loop entirely.

```r
i = 1

repeat {
    i = i+1
    if (i < 10) next
    print(i^2)
    if (i >= 13) break
}
```

## Function - Single Variable
`name = function(argument) { expression }` defines a function. Calling `name(value)` runs the expression with `argument` set to `value`, and the last evaluated expression is returned.

```r
abc = function(x) {
    x^2
}

abc(3)
abc(6)
abc(9)
```

## Function - Two Variables
A function can take multiple arguments, separated by commas, and use all of them inside its body.

```r
abc = function(x, y) {
    x^2+y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

## Function
A function body can contain any valid R expression, including calls to built-in functions like `sin()` and `cos()`.

```r
abc = function(x) {
    sin(x)^2+cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)
```

## Function Without Argument
A function can be defined with empty parentheses `function()` when it needs no input, and is called as `name()`.

```r
abc = function() {
    for (i in 1:3) {
        print(i^3)
    }
}

abc()
```

## Sequences - Basic
`seq(from, to)` generates a regular sequence of numbers between `from` and `to`, using a default step of +1 or -1 depending on direction.

```r
seq(from=2, to=4)
seq(from=4, to=2)
seq(from=-4, to=4)
```

## Sequences - Constant Increment
The `by` argument sets a custom step size, so `seq(from, to, by)` increases in steps of `by`.

```r
seq(from=10, to=20, by=2)
```

## Sequences - Constant Decrement
Using a negative `by` value produces a decreasing sequence from `from` down to `to`.

```r
seq(from=20, to=10, by=-2)
```

## Sequences - Fractional Decrement
The `by` step can be a decimal, allowing fractional increments or decrements between `from` and `to`.

```r
seq(from=3, to=-2, by=-0.5)
```

## Sequences - Predefined Length
`seq(length=n)` fixes the number of elements in the sequence instead of the step size, using the default increment of +1.

```r
seq(to=10, length=10)
seq(from=10, length=10)
```

## Sequences - Fractional Increment with Length
`length` and `by` can be combined: `length` fixes how many values are produced, and `by` sets the fractional step between them.

```r
seq(from=10, length=10, by=0.1)
```

## Sequences - Constant Decrement with Length
A negative `by` with `length` produces a fixed number of decreasing values starting from `from`.

```r
seq(from=10, length=10, by=-2)
```

## Sequences - Fractional Decrement with Length
Combining a fractional negative `by` with `length` gives a fixed-size sequence decreasing in fractional steps.

```r
seq(from=10, length=5, by=-.2)
```
