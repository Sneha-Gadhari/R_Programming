## 1. Detecting Missing Values with `is.na()`

R stores missing observations as `NA`. `is.na(x)` returns a logical vector, `TRUE` wherever a value is missing.

````r
x = NA
is.na(x)
````

## 2. `is.na()` on a Vector

`is.na()` works on vectors, lists, matrices, and data frames, returning one `TRUE`/`FALSE` per element.

````r
x = c(11, NA, 13, NA)
is.na(x)
````

## 3. Handling `NA` in `mean()`

Arithmetic functions like `mean()` return `NA` if any input is missing. Setting `na.rm = TRUE` drops the missing values before computing.

````r
x = c(11, NA, 13, NA)
mean(x)
mean(x, na.rm = TRUE)
````

## 4. `NA` versus `NULL`

`NA` is a placeholder for a value that exists but is missing. `NULL` represents something that never existed at all; it is returned by some functions and expressions.

## 5. Locating Missing Values with `which()`

`which(is.na(x))` returns the index positions where `x` is `NA`.

````r
x = c(11, NA, 13, NA)
x
which(is.na(x))
````

## 6. Counting Missing Values with `sum()`

Since `TRUE`/`FALSE` are treated as `1`/`0`, `sum(is.na(x))` counts how many values are missing.

````r
x = c(11, NA, 13, NA)
x
sum(is.na(x))
````

## 7. Finding Complete Cases

`complete.cases(x)` returns a logical vector marking the positions that are **not** missing.

````r
x = c(11, NA, 13, NA)
x
complete.cases(x)
````

## 8. Removing Missing Values with `na.omit()`

`na.omit(x)` returns the object with missing values deleted listwise, permanently dropping any row/element that contains `NA`.

````r
x = c(11, NA, 13, NA)
x
y = na.omit(x)
y
mean(x)
mean(y)
````

## 9. Conditional Execution: `if()`

```
if (condition) {
    execute commands if condition is TRUE
}
```
The code inside `{}` runs only when the condition is `TRUE`; if it is `FALSE`, R silently skips it and gives no output. `if()` should only be used with a single-element (scalar) condition, not a vector.

````r
x = 5
if (x > 4) x * 3

x = 3
if (x > 4) x * 3
````

### Same Example Using `{}`

````r
x = 5
if (x > 4) {
    x * 3
}

x = 3
if (x > 4) {
    x * 3
}
````

### Printing a Message Conditionally

````r
x = 6
if (x > 3) {
    print("The value is more than 3")
}
````

````r
x = 2
if (x > 3) {
    print("The value is more than 3")
}
````

## 10. Conditional Execution: `if...else()`

```
if (condition) {
    executes commands if condition is TRUE
} else {
    executes commands if condition is FALSE
}
```
The condition should be a single value, not a vector, and may combine expressions using `&&` (and) or `||` (or).

````r
x = 5
if (x == 3) {
    x = x - 1
} else {
    x = 2 * x
}
x
````

````r
x = 3
if (x == 3) {
    x = x - 1
} else {
    x = 2 * x
}
x
````

### Printing Which Branch Was Taken

````r
x = 6
if (x > 3) {
    print("The value is more than 3")
} else {
    print("The value is less than 3")
}
````

````r
x = 2
if (x > 3) {
    print("The value is more than 3")
} else {
    print("The value is less than 3")
}
````

## 11. Nested `if...else...if`

When there are more than two alternatives, chain `else if` blocks:
```
if (condition1) {
    ...
} else if (condition2) {
    ...
} else if (condition3) {
    ...
} else {
    executes commands if all conditions are FALSE
}
```

````r
x = 5
if (x == 3) {
    x = x - 1
} else if (x < 3) {
    x = x + 5
} else {
    x = 2 * x
}
x
````

````r
x = 2
if (x == 3) {
    x = x - 1
} else if (x < 3) {
    x = x + 5
} else {
    x = 2 * x
}
x
````

````r
x = 3
if (x == 3) {
    x = x - 1
} else if (x < 3) {
    x = x + 5
} else {
    x = 2 * x
}
x
````

## 12. Vectorized Conditional: `ifelse()`

```
ifelse(test, yes, no)
```
Unlike `if()`, `ifelse()` evaluates a condition element-by-element over an entire vector, returning `yes` where the test is `TRUE` and `no` where it is `FALSE`.

````r
x = 1:10
x
ifelse(x < 6, x^2, x + 1)
````

### Odd/Even Example (Modulo `%%`)

`%%` gives the remainder of division. A remainder of `0` when dividing by `2` means the number is even.

````r
x = c(7, 9, 8, 4)
ifelse(x %% 2 == 0, "even number", "odd number")
````

## 13. `switch()` for Multiway Branching

```
switch(expr, ...)
```
`switch` compares `expr` against a list of cases and returns the matching value. A numeric `expr` selects the corresponding position; a character `expr` is matched against named arguments. If there is no match and no unnamed fallback element, `switch` returns nothing.

````r
switch(2, "apple", "banana", "orange")
switch(1, "apple", "banana", "orange")
````

### Matching by Name

````r
switch("colour", colour = "blue", gender = "male", volume = 50)
switch("volume", colour = "blue", gender = "male", volume = 50)
````

### No Match Case

````r
switch(4, "apple", "banana", "orange")
switch("size", colour = "blue", gender = "male", volume = 50)
````

## 14. `which()` for Positions Satisfying a Condition

```
which(x, arr.ind, useNames)
```
`which()` returns the indices of a logical vector that are `TRUE`. With `arr.ind = TRUE` on a matrix, it returns row/column positions instead of flat indices.

````r
x = c(10, 15, 8, 14, 6, 12)
x
which(x == 14)
which(x != 12)
which(x > 10)
````

### `which.min()` and `which.max()`

````r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x
which.min(x)
which.max(x)
````

### `which()` with `arr.ind = TRUE` on a Matrix

````r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x
which(x %% 2 == 1)
which(x %% 2 == 1, arr.ind = TRUE)
````
