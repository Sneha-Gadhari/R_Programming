## Part 1: Strings

### 1.1 Replace first match with `sub`
`sub(old, new, string)` replaces only the first occurrence of `old` in `string`.

```r
y = "Number of participants: 25"
sub("25", "30", y)
```

### 1.2 `sub` replaces only the first instance

```r
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."
sub("Mr. Singh", "Professor Jha", y)
```

### 1.3 Replace all matches with `gsub`
`gsub(old, new, string)` performs a global replace, so every occurrence of `old` is replaced.

```r
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."
gsub("Mr. Singh", "Professor Jha", y)
```

### 1.4 Compare `sub` and `gsub`

```r
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."
gsub("Mr. Singh", "Professor Jha", y)
sub("Mr. Singh", "Professor Jha", y)
```

### 1.5 Search with `grep` (indices)
`grep(pattern, x, ignore.case = FALSE, value = FALSE)` searches each element of vector `x` for `pattern`.

- `value = FALSE` (default) returns the integer indices of matching elements.
- `value = TRUE` returns the matching elements themselves.
- `ignore.case = TRUE` ignores case during matching.

```r
str = c("R Course", "exercises", "include examples of R language")
grep("ex", str, value = FALSE)
```

### 1.6 `grep` returning matching elements

```r
str = c("R Course", "exercises", "include examples of R language")
grep("ex", str, value = TRUE)
```

### 1.7 `grep` with `ignore.case`
Matching is case-sensitive by default. Set `ignore.case = TRUE` to match both `R` and `r`.

```r
str = c("R Course", "exercises", "include examples of r language", "in R software.")
grep("R", str, ignore.case = FALSE, value = TRUE)
grep("R", str, ignore.case = TRUE, value = TRUE)
```

### 1.8 `grep` indices with `ignore.case`

```r
grep("R", str, ignore.case = TRUE, value = FALSE)
grep("R", str, ignore.case = FALSE, value = FALSE)
```

### 1.9 `grep` on combined strings
`c(x, y)` combines strings into one vector, and `grep` returns the position of each element containing the pattern.

```r
x = "R course 24.07.2022"
y = "Number of participants: 50"
c(x, y)
grep("our", c(x, y))
grep("Num", c(x, y))
```

### 1.10 Logical search with `grepl`
`grepl(pattern, x)` returns a logical vector: `TRUE` where the pattern matches, `FALSE` otherwise.

```r
str = c("R Course", "exercises", "include examples of R language")
grepl("R", str)
grepl("ex", str)
```

## Part 2: Data Frames

A data frame combines variables of equal length, with each row holding observations on the same unit. Unlike `cbind` or `matrix`, it can hold numeric, character and factor columns together.

### 2.1 Load the `painters` data frame
`painters` comes from the `MASS` package. Row names are the painters' names.

```r
library(MASS)
painters
```

### 2.2 Row names and column names
`rownames(df)` and `colnames(df)` return the row and column labels.

```r
rownames(painters)
colnames(painters)
```

### 2.3 Extract a column with `$`
`df$column` extracts one variable. Use `is.numeric` and `is.factor` to check its type.

```r
is.numeric(painters$School)
is.numeric(painters$Drawing)
is.factor(painters$School)
is.factor(painters$Drawing)
```

### 2.4 Summary of a data frame
`summary()` is generic. For numeric variables it gives descriptive measures, and for a factor it gives a frequency table.

```r
summary(painters)
summary(painters$School)
```

### 2.5 `attach` a data frame
`attach(df)` lets you use column names directly, without the `df$` prefix.

```r
attach(painters)
summary(School)
summary(Composition)
```

### 2.6 `detach` a data frame
`detach(df)` restores the default behaviour, so `df$` is required again. The error below is expected.

```r
detach(painters)
tryCatch(
    summary(School),
    error = function(e) print(conditionMessage(e))
)
```

### 2.7 Subset rows with `subset`
`subset(df, condition)` keeps rows where the condition is `TRUE`. Use `==` for equality.

```r
subset(painters, School == "F")
subset(painters, Composition <= 6)
```

### 2.8 Subset rows with square brackets
`df[condition, ]` is equivalent. The empty position after the comma means all columns.

```r
painters[painters[["School"]] == "F", ]
```

### 2.9 Drop columns with `select`
`select = c(-3, -5)` removes the third and fifth columns (Colour and School).

```r
subset(painters, School == "F", select = c(-3, -5))
```

### 2.10 Partition with `split`
`split(df, factor)` divides a data frame into a list of data frames, one per level of the factor.

```r
splitted = split(painters, painters$School)
splitted
```

### 2.11 Access one group
Each element of the list is itself a data frame.

```r
splitted$A
is.data.frame(splitted$A)
```

### 2.12 Create two data frames to combine
Both share the `state` column.

```r
df1 = data.frame(
    state = c("UP", "MP", "AP", "JK"),
    popnsize = c(1000, 2000, 3000, 4000)
)

df2 = data.frame(
    state = c("UP", "MP", "AP", "JK"),
    samplesize = c(100, 200, 300, 400),
    surveycompleted = c("Yes", "No", "Yes", "No")
)

df1
df2
```

### 2.13 Combine columns with `cbind`
`cbind(df1, df2)` places data frames side by side. Rows are matched by position, and the duplicate `state` column is kept.

```r
cbind(df1, df2)
```

### 2.14 Join on a common column with `merge`
`merge(x, y, by = "column")` joins data frames using a shared column. Rows are matched by value, not position, and the result is sorted by the key by default.

```r
merge(df1, df2, by = "state")
```

### 2.15 Create two data frames to stack
Both must have the same column names.

```r
df11 = data.frame(
    state = c("UP", "MP", "AP", "JK"),
    popnsize = c(1000, 2000, 3000, 4000)
)

df22 = data.frame(
    state = c("Bihar", "Delhi", "Punjab"),
    popnsize = c(100, 200, 300)
)

df11
df22
```

### 2.16 Stack rows with `rbind`
`rbind(df11, df22)` appends the rows of the second data frame below the first.

```r
rbind(df11, df22)
```
