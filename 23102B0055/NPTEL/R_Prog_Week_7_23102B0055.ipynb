##More sequences and other operations

## 1. Colon Operator `:`
The colon operator `a:b` generates a sequence with a constant increment of 1 (or -1 if `a > b`). It also works with decimals and negative numbers.

```r
1:10
10:1
5:15
15:5
```

```r
-1:-10
-10:-1
-5:-15
-15:-5
```

```r
1.23:10
1.23:10.54
10.54:2.23
```

## 2. The `seq()` Function
`seq(from, to, by, length.out, along.with)` generates regular sequences. `seq(10)` is same as `1:10`.

```r
seq(10)
seq(1:10)
```

## 3. `seq()` with Predefined Variable and Increment
The `by` argument can be an expression, e.g. `x/10`, to control the step size.

```r
x=2
seq(1,x,x/10)
```

```r
x=50
seq(0,x,x/10)
```

## 4. Storing Sequence Output
The output of `seq()` can be assigned to a variable and reused in further computations.

```r
x=seq(1,50,1/2)
x
```

```r
y=2*x
y
```

## 5. Index Vectors
`seq(along = x)` creates an index vector matching the length of `x`. Elements of `x` can then be accessed using this index vector.

```r
x=c(9,8,7,6)
ind = seq(along=x)
ind
```

```r
x[ind[2]]
```

## 6. Sys.time() and Sys.Date()
`Sys.time()` returns the current date and time; `Sys.Date()` returns only the current date, from the system clock.

```r
Sys.time()
```

```r
Sys.Date()
```

## 7. Sequences of Dates
`seq(from, to, by, length.out)` also works on `Date` objects. The `by` argument accepts `"day"`, `"week"`, `"month"`, `"quarter"`, or `"year"`.

```r
seq(as.Date("2010-01-01"), as.Date("2017-01-01"), by="years")
```

```r
seq(as.Date("2017-01-01"), by="days", length=6)
```

```r
seq(as.Date("2017-01-01"), by="months", length=6)
```

```r
seq(as.Date("2017-01-01"), by="years", length=6)
```

### Sequence Between Two Dates (Reverse / Negative Step)
Using a negative `by` value like `"-1 month"` generates the sequence from the end date back to the start date.

```r
startdate = as.Date("2016-01-01")
enddate = as.Date("2017-01-01")
output = seq(enddate, startdate, by = "-1 month")
output
output = seq(startdate, enddate, by="month")
output
```

## 8. Sequence of Alphabets
`letters` gives the sequence of lowercase alphabets and `LETTERS` gives uppercase alphabets. Indexing `letters[from:to]` extracts a subrange.

```r
letters
```

```r
letters[1:3]
letters[3:1]
letters[17:20]
letters[2]
```

```r
LETTERS
```

## 9. Repeats — `rep()`
`rep(x, times = n)` repeats the entire vector `x`, `n` times. `rep(x, each = n)` repeats each element `n` times before moving to the next.

```r
rep(3.5, times=10)
rep(1:4, 2)
```

```r
x=1:4
x
rep(x, times=3)
rep(x, each=3)
```

### Combining `each` and `times`
When both arguments are used together, the order in which they are specified affects the resulting sequence.

```r
rep(1:4, each=2)
rep(1:4, each=2, times=3)
rep(1:4, times=3, each=2)    #regardless of order each is first
```

### Repeating Each Element a Different Number of Times
If the second argument to `rep()` is itself a vector, each element of `x` is repeated according to the corresponding value in that vector.

```r
rep(1:4, 2:5)
```

```r
ans=seq(from=2, to=8, by=2)
ans
rep(1:4, ans)
```

### Repeating a Matrix
`rep()` applied to a matrix repeats its elements column-wise as a flat vector.

```r
x=matrix(nrow=2, ncol=2, data=1:4, byrow=T) #by default its columnwise so byrow function does rowwise
x
rep(x,3) #columnwise
```

### Repetition of Character Vectors

```r
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)
```

### Repetition for a Pre-Specified Output Length
`rep(x, length.out = n)` (or `length = n`) recycles `x` until the output has length `n`.

```r
rep(2, length.out=5)
rep(2, length=5)
rep(c(2,3), length=5)
rep(c(2,3,4), length=5)
```

```r
rep("apple", length=5)
rep(c("a","b","c"), length=5)
rep(c("a","b","c"), length=2)
```

## 10. Sorting — `sort()`
`sort(x, decreasing = FALSE)` arranges the values of `x` in ascending order by default; set `decreasing = TRUE` for descending order.

```r
y=c(8,5,7,6)
y
sort(y)
sort(y, decreasing = TRUE)
```

## 11. Ordering — `order()`
`order(x)` returns the **indices/positions** that would arrange `x` in ascending order, rather than the sorted values themselves.

```r
y=c(9,8,5,7,6)
y
order(y)
order(y, decreasing=TRUE)
```

## 12. `mode()` Function
`mode()` returns how an object is stored internally, e.g. `"numeric"`, `"character"`, `"list"`, or `"function"`.

```r
mode(2.432)
mode(c(3,4,5,6,7)) #mode of content is numeric
mode("India")
mode(c("India", "Australia")) #mode of vector is depend on its content here it is character
```

```r
mode(factor(c("UP","MP")))
mode(list("India", "USA"))
mode(data.frame(x=1:2, y=c("India", "USA")))
mode(print)
```

## 13. Lists
A list can hold elements of **different modes and types** (numbers, characters, matrices, etc.), unlike a vector where all elements must share the same mode. List elements are accessed with `[[ ]]`.

```r
x1=matrix(nrow=2, ncol=2, data=1:4, byrow=T)
x2=matrix(nrow=2, ncol=2, data=5:8, byrow=T)
x1
x2
x1 + x2
```

### Mixing Types Breaks Numeric Operations
Assigning a character value into a numeric matrix coerces the whole matrix to character mode, so arithmetic operations then fail.

```r
x1[2,1] = "hello"
x1
```

### Creating a List of Matrices

```r
matlist = list(x1, x2)
matlist
```

### Single Bracket `[ ]` vs Double Bracket `[[ ]]`
`matlist[1]` returns a **sublist** (still a list), while `matlist[[1]]` returns the actual element itself.

```r
matlist[1]
matlist[2]
```

### List with Mixed Object Types
A single list can hold a character vector, a numeric vector, and a matrix together.

```r
z1=list(c("water", "juice", "lemonade"), rep(1:4, each=2), matrix(data=5:8, nrow=2, ncol=2, byrow=T))
z1
```

### Extracting Elements from a List
`z1[1][2]` indexes the sublist and returns `NULL` since a sublist has only one element. `z1[[1]][2]` first extracts the actual vector, then indexes into it correctly.

```r
z1[[1]]
```

```r
z1[1][2] #bcoz it considers 1d array
```

```r
z1[[1]][2]
```
