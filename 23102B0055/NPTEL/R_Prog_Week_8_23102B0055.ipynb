### Operations on Lists, Vector Indexing, Factors, Strings Display and Formatting

## 1. Merging Lists
Two lists can be merged into one using the `c()` function, which concatenates the elements of both lists into a single list.

```r
list1=list(1,2,3)
list2=list("water", "juice", "lemonade")
list12=c(list1,list2)
print(list12)
```

## 2. Converting List to Vector
The `unlist()` function flattens a list into a plain vector. Use `mode()` to check the type before and after.

```r
list1=list(1,2,3)
list2=list("water", "juice", "lemonade")
print(unlist(list1))
print(unlist(list2))
print(mode(list1))
print(mode(list2))
mode(unlist(list1))
```

## 3. Appending to a List
The `append()` function adds a new element to the end of a list by default.

```r
list1=list(1,2,3)
list2=list("water", "juice", "lemonade")
print(append(list1, 100))
print(append(list2, "coffee"))
```

## 4. Appending After a Specific Position
The `after` argument of `append()` specifies the position after which the new element should be inserted.

```r
print(append(list1, 100, after=2))
print(append(list2, "coffee", after=2))
```

## 5. Removing an Element from a List
Using a negative index, `listname[-position]`, removes the element at that position and returns the rest.

```r
print(list1[-2])
print(list2[-1])
```

## 6. Extracting a Range from a List
A range of indices, `listname[start:end]`, or a set of indices using `c()` extracts multiple elements at once.

```r
list1=list(1,2,3,4,5,6)
print(list1[2:4])   #its not element
print(list1[c(1,3,5)])         #its not element
```

```r
list2=list("a","b","c","d","e","f")
print(list2[2:4])  #index
print(list2[c(1,3,5)])    #index
```

## 7. Vector Indexing with Logical Conditions
A vector can be indexed using a logical condition inside `[]`. Only elements satisfying the condition are returned. `%%` gives the modulus (remainder).

```r
x=1:10
print(x)
print(x[(x>5)])
print(x[x%%2==0])
print(x[x%%2==1])
```

## 8. Vector with Missing Observations
`NA` represents a missing value. The `is.na()` function detects missing values, and `!is.na()` filters them out before computing statistics like `mean()`.

```r
x=1:10
x[5] = NA
print(x)
y=x[is.na(x)]
print(y)
y = x[!is.na(x)]
print(y)
print(mean(x))
print(mean(y))
```

## 9. Vector Indexing with Negative Integers
A negative index range, `x[-(1:5)]`, excludes those positions, giving the same result as directly selecting the remaining positions.

```r
x=1:10
print(x)
print(x[-(1:5)]) #does not permanently remove
print(x[(6:10)])
x[-(1:5)] #does not permanently remove
print(x)
```

## 10. Named List Elements
List elements can be named at creation, for example `list(a1 = 1, a2 = "c")`. The `names()` function retrieves or sets these names.

```r
z=list(a1=1, a2="c", a3=1:3)
print(z)
print(names(z))
```

## 11. Renaming a Specific List Element
A single element's name can be changed using `names(z)[position] = "newname"`.

```r
z=list(a1=1, a2="c", a3=1:3)
names(z)[3] = "c2"     #a3 changed to c2
print(z)
```

## 12. Named Vectors
Vectors can also have named elements, and elements can then be accessed directly by name using `x["name"]`.

```r
x=c(water=1, juice=2, lemondae=3)
print(names(x))
print(x["juice"])
```

## 13. Empty Index
Using an empty index, `x[]`, returns the entire vector unchanged.

```r
x=1:10
print(x)
print(x[])
```

## 14. Mixed Mode List as a Matrix
A list containing both numbers and characters is heterogeneous. Assigning a `dim` attribute to it arranges it into a matrix-like structure while it still remains a list.

```r
ab=list(1,2,3,"x","y","z")
dim(ab)=c(2,3) #by default column wise
print(ab)
print(mode(ab))
```

## 15. Creating a Factor
`factor(x)` encodes a categorical vector as a factor. By default, levels are ordered alphabetically or numerically.

```r
x=c(2,3,4,5,1,1,2,2,3,3,3,4,4,5)
y=as.factor(x)
print(y)
```

```r
x=factor(c("lemonade", "juice", "water", "tea", "coffee", "juice")) #factor will sort levels alphabetically
print(x)
```

## 16. Factors with Custom Levels and Labels
The `levels` argument fixes the order of categories, and `labels` assigns readable names to each level.

```r
y=c(1,3,4,2,5,4,4)
possible.dieface=c(1,2,3,4,5,6)
labels.dieface=c("one","two","three","four","five","six")
facy= factor(y,labels=labels.dieface, levels=possible.dieface)
print(facy)
```

## 17. The class() Function
`class()` reports the class of any R object, such as `numeric`, `character`, `function`, `matrix`, or `factor`.

```r
print(class(9))
print(class("9"))
print(class(print))
x=matrix(nrow=2,ncol=2,data=1:4)
print(class(x))
```

## 18. The unclass() Function
`unclass()` strips the factor class, revealing the underlying integer codes and the `levels` attribute.

```r
brands=c("A","A","B","B","B","B","B","C","C")
brands_fac=factor(brands)
print(brands_fac)
print(unclass(brands_fac))
```

## 19. Using unclass() Codes to Map to Another Vector
The integer codes from `unclass()` can be used to index a separate vector, mapping each factor level to a corresponding value.

```r
colours=c("blue", "red", "green")
brands=c("A","A","B","B","B","B","C","C")
brands_fac=factor(brands)
print(colours[unclass(brands_fac)])
```

## 20. Custom Level Order in Factors
Passing a specific `levels` argument overrides the default ordering, and `levels()` retrieves the current level order.

```r
x=factor(c("lemonade","lemonade","juice","water","lemonade"), levels=c("water", "juice", "lemonade"))
print(x)
print(unclass(x))
print(levels(x))
```

## 21. Ordered Factors
The `ordered()` function creates a factor with a meaningful rank among levels, shown using `<` between level names.

```r
income=ordered(c("high","high","low","medium","medium"),levels=c("low","medium","high"))
print(income)
print(unclass(income))
```

## 22. The print() Function
`print()` displays its argument. The `digits` argument controls how many significant digits are shown for numeric values.

```r
print(sqrt(2))
print(sqrt(2), digits=5)
print(sqrt(2), digits=10)
```

## 23. Printing Character Vectors
`print()` displays character vectors with quotes, and mixing characters with numbers converts everything to character type.

```r
print("apple")
print(c("apple", "banana"))
print(c("apple","banana",6,10))
```

## 24. The format() Function
`format()` prepares an object for pretty printing. `digits` sets significant digits, and `nsmall` sets the minimum digits after the decimal point.

```r
print(format(0.5, digits=10, nsmall=17))
```

## 25. format() with width and justify
The `width` argument sets the total field width, and `justify` controls alignment: `"left"`, `"right"`, `"centre"`, or `"none"`.

```r
print(format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre"))
print(format(c("A", "BB", "CCC", "DDDD"), width = 14, justify = "centre"))
```

```r
print(format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "left"))
print(format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre"))
print(format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "right"))
print(format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "none"))
```

## 26. Formatting a Matrix for Display
A matrix printed with `print()` can also use the `digits` argument to control the displayed precision.

```r
x=matrix(nrow=3,ncol=2,data=1:6,byrow=T)
print(x)
```

## 27. Formatting Large Numbers with big.mark
The `big.mark` argument inserts a separator, such as a comma or space, between digit groups in large numbers.

```r
print(format(1234567, big.mark=","))
print(format(12345678, big.mark=","))
print(format(123456789, big.mark=","))
print(format(123456789, big.mark=" "))
```
