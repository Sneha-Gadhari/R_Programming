# R Programming - Matrix Operations & Logical/Relational Operators


## 1. Creating a Matrix and Renaming Row/Column Names
A matrix is built with `matrix(nrow, ncol, data)`. `rownames()` and `colnames()` label the rows and columns.

```r
x = matrix(nrow = 4, ncol = 3, data = c(1:12))
x

rownames(x) = c("r1", "r2", "r3", "r4")
colnames(x) = c("c1", "c2", "c3")
x
```

## 2. Assigning a Specified Number to All Matrix Elements
If `data` is a single value, it is recycled to fill every cell of the matrix.

```r
x = matrix(nrow = 4, ncol = 2, data = 2)
x
```

## 3. Diagonal Matrix
`diag(value, nrow, ncol)` creates a diagonal matrix with `value` on the diagonal and 0 elsewhere.

```r
d = diag(1, nrow = 3, ncol = 3)
d

d = diag(5, nrow = 3, ncol = 3)
d
```

## 4. Transpose of a Matrix
`t(x)` returns the transpose X' of matrix X, swapping rows and columns.

```r
x = matrix(nrow = 4, ncol = 2, data = 1:8, byrow = T)
x

xt = t(x)
xt
```

## 5. Row and Column Sums
`rowSums(x)` adds elements across each row, `colSums(x)` adds elements down each column.

```r
x = matrix(nrow = 4, ncol = 2, data = c(1,2,3,4,5,6,7,8))
x

rowSums(x)
colSums(x)
```

## 6. Row and Column Means
`rowMeans(x)` and `colMeans(x)` compute the average of each row and column respectively.

```r
x = matrix(nrow = 4, ncol = 2, data = c(1,2,3,4,5,6,7,8))
x

rowMeans(x)
colMeans(x)
```

## 7. Access to Rows, Columns or Submatrices
Use `x[row, col]` indexing. A blank index means "all". Ranges use `:` and arbitrary sets use `c()`.

```r
x = matrix(nrow = 5, ncol = 3, byrow = T, data = 1:15)
x

x[3, ]
x[, 2]
x[4:5, 2:3]
x[c(1,4), c(1,3)]
```

## 8. Scalar Addition, Subtraction, Multiplication and Division
Arithmetic between a matrix and a single number is applied element-wise to every entry.

```r
x = matrix(nrow = 4, ncol = 2, data = 1:8, byrow = T)

x + 5
x - 5
5 * x
x / 2
```

## 9. Addition and Subtraction of Matrices
Matrices of the same dimension can be added or subtracted directly using `+` and `-`, element by element.

```r
x = matrix(nrow = 4, ncol = 2, data = 1:8, byrow = T)
y = matrix(nrow = 4, ncol = 2, data = 11:18, byrow = T)

x + y
x - y
```

## 10. Multiplication of Matrices
`%*%` performs true matrix multiplication (not element-wise). Dimensions must be conformable: (n x k) %*% (k x m).

```r
x = matrix(nrow = 4, ncol = 2, data = 1:8, byrow = T)
y = matrix(nrow = 2, ncol = 4, data = 11:18, byrow = T)

x %*% y
y %*% x

t(x) %*% x
x %*% t(x)
```

## 11. Cross Product
`crossprod(x)` computes X'X faster than the equivalent `t(x) %*% x`.

```r
x = matrix(nrow = 4, ncol = 2, data = 1:8, byrow = T)

crossprod(x)
```

## 12. Concatenating Matrices
`rbind(x, y)` stacks matrices row-wise; `cbind(x, y)` joins matrices column-wise.

```r
x = matrix(nrow = 3, ncol = 2, data = 1:6, byrow = T)
y = matrix(nrow = 3, ncol = 2, data = 11:16, byrow = T)

rbind(x, y)
cbind(x, y)
```

## 13. Inverse of a Matrix
`solve(y)` returns the inverse of a square, invertible (positive definite) matrix.

```r
y = matrix(nrow = 2, ncol = 2, byrow = T, data = c(84,100,100,120))
y

solve(y)
```

## 14. Eigenvalues and Eigenvectors
`eigen(y)` returns a list with `$values` (eigenvalues) and `$vectors` (corresponding eigenvectors) of a square matrix.

```r
y = matrix(nrow = 2, ncol = 2, byrow = T, data = c(84,100,100,120))

eigen(y)
```

## 15. Relational Operators
`>`, `>=`, `<`, `<=`, `==`, `!=` compare values element-wise and return `TRUE`/`FALSE`.

```r
8 > 7
7 < 5
7 >= 7
8 <= 8
8 != 9
7 == 7
```

## 16. Logical Operators: `&` vs `&&` and `|` vs `||`
`&` and `|` compare every element of a vector (element-wise). `&&` and `||` look only at the first element of each vector and stop as soon as the result is known.

```r
x = c(8, 18)

(x < 10) & (x > 2)
(x[1] < 10) && (x[1] > 2)

(x < 10) | (x < 2)
(x[1] < 10) || (x[1] < 2)
```

## 17. Using Logical Vectors to Filter Data
A logical vector produced by comparisons can be used inside `x[...]` to extract only the elements that satisfy the condition.

```r
x = 1:6

(x > 2) & (x < 5)
x[(x > 2) & (x < 5)]

(x > 2) | (x < 5)
x[(x > 2) | (x < 5)]
```

## 18. Negation, xor, isTRUE and isFALSE
`!x` negates a logical value. `xor(x, y)` is TRUE only when exactly one of x, y is TRUE. `isTRUE()`/`isFALSE()` test a single logical value.

```r
x = TRUE
y = FALSE

!x
xor(x, y)
isTRUE(8 > 6)
isFALSE(5 > 8)
```
