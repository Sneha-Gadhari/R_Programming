# Foundations of R Software: Strings, Display and Formatting

## 1. Print and Format with `cat`

`print()` outputs only one object at a time. `cat()` converts its arguments to character vectors, joins them in sequence, and writes a continuous output.

```r
cat(..., file = "", sep = " ", fill = FALSE, labels = NULL, append = FALSE)
```

- `sep`: string placed between items (default is a single space).
- `fill`: if `TRUE` or a positive number, output is broken into lines of that width.
- `labels`: character vector of labels for the lines printed when `fill` is used.
- `append`: used only when `file` is a file name; `TRUE` appends instead of overwriting.
- `cat` does not add a newline; use `"\n"` to end a line.

### print with Multiple Items (Error)

````r
tryCatch(
    print("The zero occurs at", 2*pi, "radians."),
    error = function(e) cat("Error:", conditionMessage(e), "\n")
)
````

### print One Item at a Time

````r
print("The zero occurs at")
print(2*pi)
print("radians")
````

### cat with Multiple Items

````r
cat("The zero occurs at", 2*pi, "radians.", "\n")
````

### cat with date()

````r
d = date()
cat("Today's date is:", d, "\n")
````

### cat with sep

````r
x = 1:10
x
cat(x, sep = " ++ ")
cat("\n")
cat(x, sep = " / ")
cat("\n")
````

### cat with Expressions

````r
x = 7
cat("The square of", x, "is", x^2, "!\n")
````

### cat with format

````r
x = 7
cat("The square root of", x, "is approximately", format(sqrt(x), digits = 3), "\n")
````

### cat with a Vector

````r
evenno = c(2, 4, 6, 8, 10)
evenno
cat("The first few even numbers are:", evenno, "...\n")
````

### cat with fill and labels

````r
x = 1:10
cat(x, fill = 2, labels = paste("(", letters[1:10], "):"))
````

## 2. The `paste` Function

`paste()` converts its arguments to character strings (via `as.character`) and joins them. Unlike `cat`, the result can be assigned to a variable.

```r
paste(..., sep = " ", collapse = NULL)
paste0(..., collapse = NULL)
```

- `sep`: separator placed between the joined strings (default is a space).
- `collapse`: if given, all results are joined into a single string using this separator.
- `paste0` is `paste` with `sep = ""`.
- Vector arguments are joined element by element; shorter vectors are recycled.

### paste vs as.character

````r
paste(1:12)
as.character(1:12)
````

### paste with sep

````r
paste("Everybody", "loves", "R Programming.")
paste("Everybody", "loves", "R Programming.", sep = "*")
paste("Everybody", "loves", "R Programming.", sep = "===")
````

### paste with a Vector of Strings

````r
names = c("Prof. Singh", "Mr.  Venkat", "Dr. Jha")
names
paste(names, "is", "a good", "person.")
````

### paste with collapse

````r
names = c("Prof. Singh", "Mr.  Venkat", "Dr. Jha")
paste(names, "is", "a good", "person.", collapse = ", and ")
````

### Role of Blank Space inside Quotes

````r
paste(1, " is first", 2, " is second", 3, " is third", sep = "#")
paste(1, " is first    ", 2, " is second      ", 3, " is third    ", sep = "#")
````

### Generate Labels with paste

````r
x = paste("Ex", 1:5, sep = "_")
x
x[1]
x[2]
x[3]
x[5]
````

### paste with collapse to a Single String

````r
x = paste("Ex", 1:5, sep = "_", collapse = "")
x
````

### Difference between sep and collapse

````r
x = paste("Ex", 1:5, sep = "_")
x
x = paste("Ex", 1:5, sep = "_", collapse = "")
x
````

### paste0 and paste with a Single Vector

````r
paste0(1:10)
paste(1:10)
````

### paste0 and paste with Multiple Vectors

````r
paste0(1:10, c("st", "nd", "rd", rep("th", 7)))
paste(1:10, c("st", "nd", "rd", rep("th", 7)))
````

## 3. String Splitting with `strsplit`

`strsplit()` splits each element of a character vector into substrings. It returns a list with one element per input string.

```r
strsplit(x, split, fixed = FALSE, ...)
```

- `x`: character vector to split.
- `split`: character vector of regular expressions (or exact strings if `fixed = TRUE`) used for splitting.
- `fixed`: if `TRUE`, `split` is matched exactly; otherwise it is treated as a regular expression.
- `split = ""` splits a string into individual characters.
- Components are accessed as `result[[i]][j]`.

### Split at a Single Character

````r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"
x
strsplit(x, split = "!")
````

### Split at a Character String

````r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"
strsplit(x, split = "&!")
````

### Access Components of the Result

````r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"
y = strsplit(x, split = "!&")
y
y[[1]][1]
y[[1]][2]
y[[1]][3]
````

### Split Dates

````r
dates = c("2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27")
dates
datesplt = strsplit(dates, "-")
datesplt
````

### Convert Split Dates to a Character Matrix

````r
datemat = matrix(unlist(datesplt), nrow = 4, ncol = 3, byrow = TRUE)
datemat
````

### Convert Split Dates to a Numeric Matrix

````r
datematrix = matrix(as.numeric(unlist(datesplt)), nrow = 4, ncol = 3, byrow = TRUE)
datematrix
````

### Split a String into Characters

````r
strsplit("Shalabh", split = "")
````

## 4. Counting Characters with `nchar` and `nzchar`

```r
nchar(x, type = "chars", allowNA = FALSE, keepNA = NA)
nzchar(x, keepNA = FALSE)
```

- `nchar` returns the number of characters in each element of `x`; non-character input is coerced to character.
- `type`: one of `"bytes"`, `"chars"`, `"width"`.
- `nzchar` returns `TRUE` for non-empty strings and `FALSE` for empty strings `""`.

### nchar on Strings

````r
x = "R course 24.07.2022"
y = "Number of participants: 25"
x
y
nchar(x)
nchar(y)
````

### nzchar on Strings

````r
x = "R course 24.07.2022"
y = "Number of participants: 25"
nzchar(x)
nzchar(y)
````

### nzchar on Vectors

````r
x = c("Apple", "Banana", "Cake")
x
nzchar(x)
y = c("Apple", "", "Cake")
y
nzchar(y)
````

### nchar on Character and Numeric Vectors

````r
x = c("Apple", "Banana", "Cake")
nchar(x)
y = c(2, 4, 6)
nchar(y)
z = c(11, 222, 3333)
nchar(z)
z1 = c(1.1, 2.22, 3.333)
nchar(z1)
````

## 5. Lower and Upper Case with `tolower` and `toupper`

```r
tolower(x)
toupper(x)
```

- `tolower` converts upper-case characters to lower case.
- `toupper` converts lower-case characters to upper case.
- Non-alphabetic characters are left unchanged.

### toupper

````r
x = "R course will start from 24.07.2022"
x
toupper(x)
````

### tolower

````r
z = "INDIAN INSTITUTE OF TECHNOLOGY"
z
tolower(z)
````
