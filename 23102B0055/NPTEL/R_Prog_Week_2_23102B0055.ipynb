## R as a Calculator

R evaluates arithmetic expressions directly at the prompt and prints the result prefixed with `[1]`, the index of the first element.

```r
2 + 3
2 * 3
2 - 3
3 / 2
```

## BODMAS Rule

R follows BODMAS (Brackets, Orders, Division, Multiplication, Addition, Subtraction), evaluated left to right within the same precedence level. Only round brackets `()` are valid for grouping; `{}` and `[]` are not used for this purpose.

```r
2 * 3 - 4 + 5 / 6
(2 + 3) * 5 + 5 - 10
(((2 + 3) * 5 + 5) - 10) / 2
```

## Blank Spaces in Expressions

Whitespace around operators has no effect on the result of a calculation.

```r
2 + 5
2   +   5
2     +5
```

## Addition with Data Vectors

`c()` creates a vector. When two vectors of equal length are added, R adds them element by element.

```r
c(2, 3, 5, 7) + c(-2, -3, -5, 8)
```

## Vector Recycling in Addition

If the vectors are of unequal length, R recycles the shorter vector to match the longer one. If the longer length is not an exact multiple of the shorter length, R still computes the result but issues a warning.

```r
c(2, 3, 5, 7) + c(8, 9)
c(2, 3, 5, 7) + c(8, 9, 10)
```

## Subtraction with Data Vectors

Subtraction between vectors also works element-wise, following the same recycling rule as addition.

```r
c(2, 3, 5, 7) - c(-2, -3, -5, 8)
c(12, 13, 15, 17) - c(8, 9)
c(12, 13, 15, 17) - c(8, 9, 10)
```

## Multiplication with Data Vectors

The `*` operator multiplies vectors element-wise, again applying recycling for unequal lengths.

```r
c(2, 3, 5, 7) * c(-2, -3, -5, 8)
c(2, 3, 5, 7) * c(8, 9)
c(2, 3, 5, 7) * c(8, 9, 10)
```

## Division with Data Vectors

The `/` operator divides vectors element-wise, applying the same recycling behavior.

```r
c(24, 20, 8, 16) / c(3, 4, 2, 8)
c(24, 20, 8, 16) / c(4, 2)
c(24, 20, 8, 16) / c(4, 2, 8)
```

## Arithmetic Between a Vector and a Scalar

When one operand is a single number (scalar), R recycles it against every element of the vector.

```r
c(2, 3, 5, 7) + 10
c(12, 13, 15, 17) - 10
c(2, 3, 5, 7) * 10
c(12, 13, 15, 17) / 10
```

## Assignment Operator

Values are assigned to variables using `<-` or `=`. Historically only `<-` was available in R.

```r
x <- 20
x

x = 20
x

y = x * 2
y

z = x + y
z
```

## Assigning Numbers and Characters

Numeric values are assigned directly. Character (string) values must be enclosed in single or double quotes; an unquoted word is treated as an object name and causes an error if it does not exist.

```r
x <- 20
x = "apple"
x
x <- 'apple'
x
```

## Checking Data Type: is.numeric() and is.character()

`is.numeric()` and `is.character()` return `TRUE` or `FALSE` depending on whether the object holds a number or a character string.

```r
x = 20
is.numeric(x)
is.character(x)

y = "apple"
is.character(y)
is.numeric(y)
```

## Converting Between Numbers and Characters

`as.character()` converts a value to a string, and `as.numeric()` converts a value to a number. Converting a non-numeric string to a number produces `NA` with a coercion warning.

```r
x = 20
y = as.character(x)
is.numeric(y)
is.character(y)
y
```

```r
y = "apple"
z = as.numeric(y)
is.numeric(z)
is.character(z)
z
```

## Comment Operator

The `#` symbol marks the start of a comment. Everything from `#` to the end of the line is ignored by R.

```r
x <- 20
x
```

## Case Sensitivity

R distinguishes between uppercase and lowercase names, so `X` and `x` refer to two different objects.

```r
x <- 20
x
X <- 30
X
x
```

## Combining Values into a Vector

The `c()` function combines individual values into a single vector. Writing values separated by commas without `c()` results in a syntax error.

```r
y = c(1, 2, 3, 4, 5)
y
```

## Mode of an Object

`mode()` reports the general data type of an object, such as `"numeric"` or `"character"`. Note this is unrelated to the statistical mode (most frequent value).

```r
x = 6
mode(x)

y = "apple"
mode(y)
```

## Storage Mode

`storage.mode()` reports the internal storage type of an object, which can differ from the value returned by `mode()` (for example, a numeric value is stored as `"double"`). This is mainly relevant when interfacing with routines written in other languages such as C or FORTRAN.

```r
x = 6
storage.mode(x)

x = TRUE
storage.mode(x)

x = "apple"
storage.mode(x)
```

## Infinity

Division by zero produces `Inf` in R rather than an error. `is.finite()` and `is.infinite()` test whether a value is finite or infinite.

```r
3 / 0
5 + Inf

x = 5 + Inf
is.finite(x)
is.infinite(x)
```
