# R Programming — Week 3 Notes
### Basics of Calculations, Built-in Functions, and Matrices

## Vector Addition
Arithmetic operators like `+` work **element-wise** on vectors created with `c()`.

```r
c(2, 3, 5, 7) + c(-2, -3, -5, 8)
```

## Power Operator
`^` and `**` both raise a number to a power.

```r
2^3
2**3
```

## Power Operator with Negative and Fractional Exponents
Fractional exponents give roots, negative exponents give reciprocals.

```r
2^0.5
2**0.5
2^-0.5
```

## Power Operator Applied to a Vector (Scalar Exponent)
When a vector is raised to a scalar power, the power is applied to every element.

```r
c(2, 3, 5, 7)^2
```

## Power Operator with Two Equal-Length Vectors
When both base and exponent are vectors of the same length, the operation is element-wise, pairing corresponding elements.

```r
c(2, 3, 5, 7)^c(2, 3)
```

## Power Operator with Recycling
If the exponent vector is shorter, it is **recycled** to match the length of the base vector.

```r
c(1, 2, 3, 4, 5, 6)^c(2, 3, 4)
```

## Power Operator — Recycling Warning
If the shorter vector's length does not evenly divide the longer vector's length, R still recycles but issues a warning.

```r
c(2, 3, 5, 7)^c(2, 3, 4)
```

## Integer Division with a Scalar
Operator `%/%` performs division and discards the fractional remainder.

```r
2 %/% 2
5 %/% 2
7 %/% 3
```

## Integer Division on a Vector
The `%/%` operator applies to each element of a vector against a scalar.

```r
c(2, 3, 5, 7) %/% 2
```

## Integer Division with Two Vectors
With equal-length vectors, elements are paired and divided one-to-one.

```r
c(2, 3, 5, 7) %/% c(2, 3)
```

## Integer Division with Recycling Warning
When lengths are unequal and not multiples, R recycles the shorter vector and warns about it.

```r
c(2, 3, 5) %/% c(2, 3)
```

## Modulo Division with Scalars
Operator `%%` returns the remainder after division (x mod y).

```r
2 %% 2
3 %% 2
7 %% 3
7 %% 4
```

## Modulo Division on a Vector
`%%` applied element-wise to a vector against a scalar.

```r
c(2, 3, 5, 7) %% 2
```

## Modulo Division with Two Vectors
Equal-length vectors pair elements for the modulo operation.

```r
c(2, 3, 5, 7) %% c(2, 3)
```

## Modulo Division with Recycling Warning
Unequal, non-multiple lengths trigger a recycling warning.

```r
c(2, 3, 5) %% c(2, 3)
```

## Maximum
`max()` returns the largest value. Note: passing values directly vs. wrapping them in `c()` can behave differently in older R versions.

```r
max(1.2, 3.4, -7.8)
max(c(1.2, 3.4, -7.8))
```

## Minimum
`min()` returns the smallest value among its arguments.

```r
min(1.2, 3.4, -7.8)
min(c(1.2, 3.4, -7.8))
```

## Arithmetic Mean
`mean()` computes the average. Always pass multiple values as a single vector using `c()` for correct results.

```r
mean(2, 3, 4)
mean(c(2, 3, 4))
```

## Absolute Value
`abs()` returns the non-negative magnitude of each element.

```r
abs(-4)
abs(c(-1, -2, -3, 4, 5))
```

## Square Root
`sqrt()` computes the square root of each element.

```r
sqrt(4)
sqrt(c(4, 9, 16, 25))
```

## Sum and Product
`sum()` adds all elements; `prod()` multiplies all elements.

```r
sum(c(2, 3, 5, 7))
prod(c(2, 3, 5, 7))
```

## Rounding
`round()` rounds a value to the nearest integer (or specified decimal places).

```r
round(1.23)
round(1.83)
```

## Natural Logarithm
`log()` computes the natural log (base e) by default.

```r
log(10)
log(exp(1))
log(c(10, 100, 1000))
```

## Logarithm Base 10
`log10()` computes the log with base 10.

```r
log10(10)
log10(100)
log10(c(10, 100, 1000))
```

## Assignments
The `=` (or `<-`) operator stores a value or expression result in a variable for later use.

```r
x1 = c(1, 2, 3, 4)
x1

x2 = x1^2
x2
```

## Combining Scalars, Vectors, and Functions
Expressions can freely mix vectors, scalars, and built-in functions.

```r
c(1, 2, 3, 4) + sum(c(1, 2, 3, 4)) * prod(c(1, 2))

abs(c(1, 2, 3, 4) - sum(c(1, 2, 3, 4)) * prod(c(1, 2)))
```

## Creating a Matrix (Column-wise)
`matrix(nrow, ncol, data)` builds a matrix, filling values column by column by default.

```r
x = matrix(nrow = 4, ncol = 2, data = c(1, 2, 3, 4, 5, 6, 7, 8))
x
```

## Accessing a Matrix Element
Use `x[i, j]` to access the element in row `i`, column `j`.

```r
x[3, 2]
```

## Filling a Matrix Row-wise
Setting `byrow = TRUE` fills the matrix row by row instead of column by column.

```r
y = matrix(nrow = 4, ncol = 2, data = c(1, 2, 3, 4, 5, 6, 7, 8), byrow = TRUE)
y
```

## Matrix Properties
`dim()`, `nrow()`, and `ncol()` return the matrix dimensions.

```r
dim(x)
nrow(x)
ncol(x)
```

## Matrix Mode and Attributes
`mode()` shows the storage type; `attributes()` lists all attributes such as dimensions.

```r
mode(x)
attributes(x)
```
