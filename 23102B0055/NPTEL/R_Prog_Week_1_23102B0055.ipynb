## What is R?

R is an environment for statistical computing, data analysis, and graphics. It is free, open-source, interpreted, and works on Windows, Linux, and macOS. It was created by Ross Ihaka and Robert Gentleman.

### Check R Version

```r
R.version.string
```

## Getting Help in R

`?functionname` opens the help page for a function. `help.search("topic")` searches help pages by keyword. `help()` and `help.start()` open general help interfaces.

### Help for a Specific Function

```r
?read.table
```

### Search Help by Topic

```r
help.search("data input")
```

### General Help Interface

```r
help()
help.start()
```

## find() and apropos()

`find("name")` tells which package an object belongs to. `apropos("pattern")` returns all object names in the search list matching a pattern.

### Find Which Package Contains a Function

```r
find("lowess")
```

### List Objects Matching a Pattern

```r
apropos("lm")
```

## Worked Examples and Demonstrations

`example(functionname)` runs the built-in example for a function. `demo(topic)` runs a demonstration script showing R capabilities.

### Run a Function's Example

```r
example(lm)
```

### Run a Graphics Demo

```r
demo(graphics)
```

## Packages and Libraries in R

A package is a bundle of functions and datasets extending R's base functionality. `library(pkgname)` loads an installed package into the session.

### Load a Library

```r
library(MASS)
```

## Installing, Describing, and Exploring Packages

`install.packages("pkg")` downloads and installs a package from CRAN. `packageDescription("pkg")` shows its metadata. `library(help=pkg)` lists all functions and datasets in it.

### Install a Package

```r
install.packages("boot")
```

### View Package Description

```r
packageDescription("boot")
```

### View Package Contents

```r
library(help=boot)
```

## Managing Installed Packages

`installed.packages()` lists all installed packages. `update.packages("pkg")` updates a package. `remove.packages("pkg")` uninstalls it. `detach("package:pkg", unload=TRUE)` unloads it from the session.

### List Installed Packages

```r
head(installed.packages())
```

### Update a Package

```r
update.packages("boot")
```

### Unload a Package

```r
library(boot)
detach("package:boot", unload=TRUE)
```

### Remove a Package

```r
remove.packages("boot")
```

## Command Line versus Scripts

R commands can be typed directly at the console prompt one at a time, or written as a multi-line script and executed together. In RGui this is done via File > New script; in RStudio, commands are typed in the script editor and run with Ctrl+Enter.

### Example: Single Line Command

```r
x <- 5
x
```

### Example: Multi-Line Script

```r
x <- seq(-10, 10, length.out = 50)
y <- x
z <- outer(x, y, function(x, y) sin(sqrt(x^2 + y^2)))
persp(x, y, z, theta = 30, phi = 30, expand = 0.5, col = "lightblue")
```
